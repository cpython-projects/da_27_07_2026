**Задача 1.** Дано рядок з даними, розділеними різними символами (комами, крапками, крапками з комою). Розбийте рядок на частини.

- Вхід: `"яблуко,груша.банан;апельсин"`
- Результат: `["яблуко", "груша", "банан", "апельсин"]`

```python
import string
```

```python
string.punctuation
```

```python
text = "яблуко,груша.банан;апельсин"

for item in string.punctuation:
  text = text.replace(item, ' ')
print(text)
```

```python
text.split()
```

**Задача 4.** Дано рядок зі словами. Підрахуйте, скільки разів зустрічається кожне слово.

- Вхід: `"кіт пес кіт риба пес кіт"`
- Результат: "кіт": 3, "пес": 2, "риба": 1

```python
text = "кіт пес кіт риба пес кіт".split()
print(text)
```

```python
res = []


for word in text:
  if word not in res:
    print(word, text.count(word))
    res.append(word)
```

**Задача 5.** Дано список рядків. Об'єднайте їх в один рядок через пробіл, але лише ті, що починаються з великої літери.

- Вхід: `["яблуко", "Банан", "груша", "Апельсин"]`
- Результат: `"Банан Апельсин"`

```python
text = ["яблуко", "Банан", "груша", "Апельсин"]
res = []

for item in text:
  if item.istitle():
    res.append(item)

print(res)
```

```python
' '.join(res)
```

**Задача 6.** Дано рядок з текстом. Переверніть кожне слово окремо, порядок слів збережіть.

- Вхід: `"привіт світ Python"`
- Результат: `"тівирп тівс nohtyP"`

```python
text = "привіт світ Python".split()
res = []

for item in text:
  res.append(item[::-1])

print(res)
```

**Задача 7.** Дано два списки слів. Знайдіть слова, які є в обох списках.

- Вхід: `["кіт", "пес", "птах", "риба"]` та `["риба", "кіт", "вовк"]`
- Результат: `["кіт", "риба"]`

```python
list_1 = ["кіт", "пес", "птах", "риба"]
list_2 = ["риба", "кіт", "вовк"]

res = []

for item in list_1:
  if item not in res and item in list_2:
    res.append(item)

print(res)
```
