```python
data = [-10, 23, 0, 10, 12, 23, 34, 45, 56, 67, 76, -45, -100]
```

```python
def even_numbers(data):
  result = []
  for item in data:
    if item % 2 == 0:
      result.append(item)
  return result

def odd_numbers(data):
  result = []
  for item in data:
    if item % 2:
      result.append(item)
  return result


print(even_numbers(data))
print(odd_numbers(data))
```

```python
def even(n):
  return n % 2 == 0

def odd(n):
  return n % 2

def neg(n):
  return n < 0


def filter_numbers(data, func):
  result = []
  for item in data:
    if func(item):
      result.append(item)
  return result


print(filter_numbers(data, even))
print(filter_numbers(data, odd))
print(filter_numbers(data, neg))
```

```python
def filter_numbers(data, func):
  result = []
  for item in data:
    if func(item):
      result.append(item)
  return result


print(filter_numbers(data, lambda n: n % 2 == 0))
print(filter_numbers(data, lambda n: n % 2))
print(filter_numbers(data, lambda n: n < 0))
```
