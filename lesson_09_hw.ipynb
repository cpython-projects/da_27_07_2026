Task 1. Визначте безліч унікальних товарів, які були куплені.

````python
res = set()
for item in customer_purchases:
  product = item['Product']
  res.add(product)

print(res)
````

````python
res = {item['Product'] for item in customer_purchases}
print(res)
````

Task 2. Виведіть список унікальних Customer_ID, які купили хоча б один товар із категорії Electronics.

````python
res = {item['Customer_ID'] for item in customer_purchases if item['Category'].lower().strip() == 'electronics'}
print(res)
````

Task 3. Визначте клієнтів, які купували як в категорії Electronics, так і в категорії Furniture.

````python
users_electronucs = {item['Customer_ID'] for item in customer_purchases if item['Category'].lower().strip() == 'electronics'}
users_furniture = {item['Customer_ID'] for item in customer_purchases if item['Category'].lower().strip() == 'furniture'}

print(users_electronucs)
print(users_furniture)
````

````python
users_electronucs & users_furniture
````

Task 4. Знайдіть клієнтів, які купували товари лише з однієї категорії.

````python
from collections import defaultdict

users = defaultdict(set)

for item in customer_purchases:
  users[item['Customer_ID']].add(item['Category'])

print(users)


for user, value in users.items():
  if len(value) == 1:
    print(user)
````

Task 5. Визначте, яка категорія товарів найчастіше зустрічається.

````python
from collections import Counter

res = Counter([item['Category'].lower().strip() for item in customer_purchases])
print(res.most_common(1))
````

Task 6. Виберіть двох клієнтів і знайдіть перетин їх покупок (які товари вони купили однакові).

````python
user_1 = {item['Product'].lower().strip() for item in customer_purchases if item['Customer_ID'] == 1001}
user_2 = {item['Product'].lower().strip() for item in customer_purchases if item['Customer_ID'] == 1003}

print(user_1 & user_2)
````

Task 7. Для кожного клієнта визначте товари, які він купив, але ніхто інший із покупців не купував.

````python
users = {item['Customer_ID'] for item in customer_purchases}

for user in users:
  user_products = set()
  others_products = set()
  for item in customer_purchases:
    if item['Customer_ID'] == user:
      user_products.add(item['Product'])
    else:
      others_products.add(item['Product'])
  print(user)
  print(user_products)
  print(others_products)
  print(user_products - others_products)
  print('-' * 20)
````
