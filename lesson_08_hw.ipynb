**Task 4.** Дано список клієнтів, кожен клієнт представлений словником з полями: `"id"`, `"ім'я"`, `"вік"`, `"місто"`.

```python
[
    {"id": 1, "ім'я": "Анна", "вік": 25, "місто": "Берлін"},
    {"id": 2, "ім'я": "Іван", "вік": 30, "місто": "Берлін"},
    {"id": 3, "ім'я": "Ольга", "вік": 22, "місто": "Відень"}
]
```

Напишіть скрипт, який формує словник, у якому ключі — міста, а значення — клієнти (список) з цих міст.

````python
users = [
    {"id": 1, "ім'я": "Анна", "вік": 25, "місто": "Берлін"},
    {"id": 2, "ім'я": "Іван", "вік": 30, "місто": "Берлін"},
    {"id": 3, "ім'я": "Ольга", "вік": 22, "місто": "Відень"}
]

grouped_users = {}

for item in users:
  city = item['місто']
  if city not in grouped_users:
    grouped_users[city] = []
  grouped_users[city].append(item['id'])

print(grouped_users)
````

````python
from collections import defaultdict

users = [
    {"id": 1, "ім'я": "Анна", "вік": 25, "місто": "Берлін"},
    {"id": 2, "ім'я": "Іван", "вік": 30, "місто": "Берлін"},
    {"id": 3, "ім'я": "Ольга", "вік": 22, "місто": "Відень"}
]

grouped_users = defaultdict(list)

for item in users:
  city = item['місто']
  grouped_users[city].append(item['id'])

print(grouped_users)
````

**Task 5.** Дано список працівників, де кожен працівник представлений словником з полями `"ім'я"`, `"відділ"`.

```python
[
    {"ім'я": "Анна", "відділ": "Маркетинг"},
    {"ім'я": "Іван", "відділ": "Продажі"},
    {"ім'я": "Ольга", "відділ": "Маркетинг"}
]
```

Створіть скрипт, який групує працівників за відділами.

````python
from collections import defaultdict

users = [
    {"ім'я": "Анна", "відділ": "Маркетинг"},
    {"ім'я": "Іван", "відділ": "Продажі"},
    {"ім'я": "Ольга", "відділ": "Маркетинг"}
]

departments = defaultdict(list)
for user in users:
  departments[user['відділ']].append(user["ім'я"])
print(departments)
````

**Task 6.** Дано словник з даними щодо продажів, де ключі — дні місяця, а значення — сума продажу.

```python
{
    1: 500, 2: 700, 3: 300, 4: 800, 5: 200
}
```

Знайдіть дні з найбільшими відхиленнями від середнього значення.

````python
import statistics
import math

data = {
    1: 500, 2: 700, 3: 300, 4: 800, 5: 200
}

avg = statistics.mean(data.values())

data_diff = {k: abs(v - avg) for k, v in data.items()}

print(data_diff)
max_value = max(data_diff.values())
for k, v in data_diff.items():
  if math.isclose(max_value, v):
    print(k)
````

````python
data = [
    '1 Bob Simson 19.58$ decorations',
    '2 Mary 66.7$ food',
    '3 Mary 98.91$ toys',
    '4 Aleksa 72.29$ drinks',
    '5 Maria Simson 84.48$ food',
    '6 Aleksa 100.41$ accessories',
    '7 Mary 19.9$ accessories',
    '8 Bob Simson 83.88$ drinks',
    '9 Bob Simson 58.21$ instruments',
    '10 Maria Simson 20.61$ accessories',
]
````

````python
res = {}
for item in data:
  *_, amount, category = item.lower().split()
  amount = float(amount.replace('$', ''))
  res[category] = res.get(category, 0) + amount

print(*res.items(), sep='\n')
````
