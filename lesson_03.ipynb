# Логічні оператори: and, or, not  
У Python є три логічні (булеві) оператори, які працюють з булевими значеннями `True` і `False`:
- **`and`** — логічне «І». Результат `True`, тільки якщо **обидва** операнди `True`
- **`or`** — логічне «АБО». Результат `True`, якщо **хоча б один** з операндів `True`
- **`not`** — логічне заперечення (інверсія). Перетворює `True` на `False` і навпаки  
Ці оператори часто використовують в умовах (`if`), щоб перевіряти декілька умов одночасно

````python
x1 = True
x2 = True

x1 and x2
````

**Пояснення:** `x1 = True`, `x2 = True`. Оскільки обидві змінні `True`, оператор `and` поверне `True`.  
Правило `and`:
```True  and True   ->  True
True  and False  ->  False
False and True   ->  False
False and False  ->  False

````python
x1 = True
x2 = True

x1 or x2
````

**Пояснення:** оператор `or` повертає `True`, якщо хоча б одна з умов істинна.  
Правило `or`:
```True  or True   ->  True
True  or False  ->  True
False or True   ->  True
False or False  ->  False

### Оператор `not`
`not` — унарний оператор (застосовується до одного значення). Він інвертує булеве значення:
- `not True` → `False`
- `not False` → `True`

````python
not True
````

````python
not False
````

# Truth Test (перевірка на істинність)
У Python будь-яке значення можна перевірити на істинність за допомогою функції `bool()`. Кожне значення трактується або як **truthy** (рівнозначне `True`), або як **falsy** (рівнозначне `False`). Це лежить в основі того, як працює `if значення:`
 — Python неявно викликає `bool(значення)`.

````python
x = 5
````

Присвоюємо змінній `x` число `5`. Далі перевіримо, чи є воно truthy.

````python
bool(x)
````

**Результат:** `True` — будь-яке ненульове число вважається truthy (істинним).

````python
x = 0
bool(x)
````

**Результат:** `False` — число `0` вважається falsy (хибним). Це стосується `0`, `0.0` і `0j`.

````python
x = 'Hello'
bool(x)
````

**Результат:** `True` — непорожній рядок є truthy.

````python
x = ''
bool(x)
````

**Результат:** `False` — порожній рядок `''` є falsy.

False - це:


*   числа - 0, 0.0, 0j, False
*   послідовності - '', [], (), {}, де 0 елементів
*   None

## Умовний оператор if / elif / else  
Далі — приклади використання `if`, `elif`, `else` на практичних задачах: розрахунок знижки залежно від суми покупки.

````python
amount = 200
discount = 0
if amount > 1000:
    discount = 0.1

total = amount * (1 - discount)
print(total)
````

**Приклад 1: проста умова `if`**  
- Якщо сума (`amount`) більша за `1000`, знижка `discount = 0.1` (10%).
- Якщо умова не виконується — знижка залишається `0` (значення за замовчуванням).
- `amount = 200`, тому умова `amount > 1000` — `False`, знижки не буде, `total = 200`.

````python
amount = 2000

if amount > 1000:
    discount = 0.1
else:
    discount = 0.05


total = amount * (1 - discount)
print(total)
````

**Приклад 2: `if / else`**  
Тепер є дві гілки:
- якщо `amount > 1000` → знижка `10%`
- інакше (`else`) → знижка `5%`  
`amount = 2000`, тому спрацьовує перша гілка: `discount = 0.1`, `total = 1800.0`.

````python
amount = 300

# 1000 - 1, 2000 - 5, 10000 - 10, 15000 - 20


if amount > 15000:
    discount = 0.2
elif amount > 10_000:
    discount = 0.1
elif amount > 2_000:
    discount = 0.05
elif amount > 1_000:
    discount = 0.01
else:
    discount = 0

total = amount * (1 - discount)
print(total)

if amount <= 1000:
    discount = 0
elif amount <= 2_000:
    discount = 0.01
elif amount <= 10_000:
    discount = 0.05
elif amount <= 20_000:
    discount = 0.1
else:
    discount = 0.2

total = amount * (1 - discount)
print(total)
````

**Приклад 3: `if / elif / elif / ... / else`**  
Умови таблиці знижок: `1000 → 1%`, `2000 → 5%`, `10000 → 10%`, `15000 → 20%`.Показано **два еквівалентні способи** написати одну й ту саму логіку:  
1. Перевірка «зверху вниз» через `>` (від найбільшого порогу до найменшого) — важливо перевіряти умови саме в такому порядку, бо `elif` виконує лише першу умову, що збіглася.  
2. Перевірка через `<=` (від найменшого порогу до найбільшого) — логічно дзеркальний варіант.Обидва варіанти для `amount = 300` дають однаковий результат: знижки немає (`discount = 0`), `total = 300`.

## Порівняння з `input()` та ланцюжки умов  
Наступні приклади демонструють визначення оцінки (`grade`) за кількістю балів (`points`), введених користувачем через `input()`.

## Тернарний (умовний) вираз  
Python дозволяє записати просту конструкцію `if / else` в **одному рядку** — це називається тернарним оператором:
```python
результат = значення_якщо_true if умова else значення_якщо_false


````python
n = 6

if n % 2 == 0:
    res = 'Even'
else:
    res = 'Odd'

print(res)


res = 'Even' if n % 2 == 0 else 'Odd'
print(res)
````

**Пояснення:** обидва варіанти коду роблять одне й те саме — визначають, парне число `n` чи непарне, і виводять результат. Другий варіант (`res = 'Even' if n % 2 == 0 else 'Odd'`) — компактніший запис того ж `if/else`, зручний для простих умов в один рядок.
