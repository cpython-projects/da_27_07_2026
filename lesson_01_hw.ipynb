## 2. Частка "так" у опитуванні

Ввести загальну кількість респондентів і кількість тих, хто відповів "так". Порахувати частку у відсотках.

## 3. Індекс маси тіла (ІМТ)

Ввести зріст (у метрах) і вагу (у кг). Порахувати ІМТ = вага / зріст².
  
Приклад:  
```
Зріст (м)>>1.75
```

Ввести загальну кількість секунд. Порахувати, скільки це повних хвилин і скільки секунд залишку (`//` та `%`).

## 6. Складний відсоток (депозит)

Ввести суму вкладу, річну ставку (у %) та кількість років. Порахувати кінцеву суму за формулою складного відсотка:  
`S = P * (1 + r/100) ** n`.
  
Приклад:  
```
Сума вкладу>>10000
Ставка (%)>>7
Кількість років>>5
Кінцева сума: 14025.52
```

````python
P = float(input('Сума вкладу>>'))
r = float(input('Ставка (%)'))
n = int(input('Кількість років>>'))

S = P * (1 + r/100) ** n
print(f'{S:.2f}')
````
