### Task 2

Аналітик підраховує витрати на зберігання контейнерів на складі:

* На склад надійшло `137` контейнерів.
* Один стандартний майданчик вміщує `10` контейнерів.

1. Скільки **повних** майданчиків буде заповнено?
2. За скільки майданчиків **доведеться заплатити**, якщо орендувати можна лише цілі майданчики?
3. Скільки контейнерів залишиться на останньому неповному майданчику?

```python
import math

containers = 137
platform = 10

print(f'Full Platforms: {containers // platform}')
print(f'Platforms For Pay: {math.ceil(containers / platform)}')
print(f'Rem: {containers % platform}')
```

### Task 3

Модель прогнозування продажів спрогнозувала виторг `predicted = 14500` грн, а фактичний виторг склав `actual = 15230` грн.

1. Обчисліть абсолютну помилку прогнозу.
2. Перевірте за допомогою логічного виразу, чи входить ця помилка в припустимий ліміт у `5%` від фактичного виторгу.

```python
predicted = 14500
actual = 15230
```

```python
error = abs(predicted - actual)
print(error)
```

```python
limit = actual * 0.05
print(error <= limit)
```

### Task 7

Фінансовий аналітик розраховує суму комісії сервісу еквайрингу.
Базова сума: `10000.50` грн. Комісія: `1.75%`.

1. Створіть змінні через `Decimal` (передаючи рядки).
2. Обчисліть точну суму комісії.
3. Округліть суму комісії до копійок (2 знаки після коми), використовуючи `ROUND_HALF_UP`.

```python
from decimal import Decimal, ROUND_HALF_UP
```

```python
base = Decimal('10000.5')
rate = Decimal('0.0175')

comission = base * rate
print(comission)
comission = comission.quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)
print(comission)
```

### Task 8

Банк нараховує відсотки на депозит зі ставкою `3.5%` на суму `1000.005` у.о.

1. Порахуйте підсумкову суму з банківським округленням (до найближчого парного числа за допомогою `ROUND_HALF_EVEN`).
2. Порівняйте результат зі стандартним фінансовим округленням `ROUND_HALF_UP`.

> **Питання:** У чому різниця і чому банки віддають перевагу банківському округленню на великих масивах даних?

```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_EVEN

amount = Decimal('1000.005')
rate = Decimal('0.035')

total = amount * (rate + Decimal('1'))
print(total)
```

```python
total.quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)
```

```python
total.quantize(Decimal('0.01'), rounding=ROUND_HALF_EVEN)
```
