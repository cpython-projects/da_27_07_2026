**Завдання 1.**  

Знайти всі встановлення з `cpi_uah > 30` і платформою **iOS**

```python
q("""
SELECT device_code, first_seen, cpi_uah, os_type
FROM devices
WHERE cpi_uah > 30 AND os_type = 'iOS'
ORDER BY cpi_uah
""")
```

**Завдання 2.**  
Знайти **5 найдорожчих** встановлень (найвищий `cpi_uah`)

```python
q("""
SELECT device_code, first_seen, cpi_uah, os_type
FROM devices
ORDER BY cpi_uah DESC NULLS LAST
LIMIT 5
""")
```
