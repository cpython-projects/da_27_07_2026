## 1. Введення та виведення даних
Першим кроком у роботі аналітика є взаємодія з даними через консоль.

*   **Функція `print()`**: Використовується для виведення результатів на екран.
    *   `sep`: параметр, що задає розділювач між значеннями (за замовчуванням — пробіл).
    *   `end`: параметр, що визначає, чим закінчується рядок (за замовчуванням — перенесення на новий рядок `\n`).
*   **Функція `input()`**: Дозволяє отримувати дані від користувача.
    *   **Важливо:** `input()` завжди повертає дані у вигляді **текстового рядка (`str`)**.
    *   Щоб виконувати математичні операції, текст потрібно перетворити на числа за допомогою функцій `int()` (ціле число) або `float()` (дробове число).

```python
print(123, 45.545, sep=',', end='******')
print('Hello', "world")
```


## 2. Базові типи даних та арифметика
Аналітики найчастіше працюють з числами:
*   `int`: Цілі числа (наприклад, кількість замовлень).
*   `float`: Дробові числа (наприклад, ціна або сума податку).

### Арифметичні операції:
1.  `+`, `-`, `*`: Додавання, віднімання та множення.
2.  `/`: Звичайне ділення (результат завжди дробове число).
3.  `//`: **Цілочисельне ділення**. Повертає цілу частину, округлюючи результат у меншу сторону.
4.  `%`: **Остача від ділення**. Допомагає визначити залишок після розподілу.
5.  `**`: Піднесення до степеня. Також використовується для вилучення кореня через ступінь `0.5`.

```python
5 + 4
```

```python
5 - 4
```

```python
6 * 7
```

```python
5 / 3
```

```python
5 // 3
```

```python
5 // -3
```

```python
5 % 3
```

## 3. Задачі

```python
# Податки та націнки
price = float(input("Введіть вартість товару: "))

# Розрахунок податку (20%)
tax = price * 0.2 # Коефіцієнт 0.2

# Фінальна ціна з націнкою (наприклад, 25%)
final_price = price * 1.25 # Коефіцієнт 1.25

print(f"Сума податку: {tax}, Ціна після націнки: {final_price}")
```

```python
# Розподіл ресурсів
total_items = int(input("Загальна кількість одиниць товару: "))
container_capacity = int(input("Місткість одного контейнера: "))

# Скільки повних контейнерів отримаємо
full_containers = total_items // container_capacity # Цілочисельне ділення

# Скільки штук залишиться поза контейнерами
leftover_items = total_items % container_capacity # Остача

print(f"Повних контейнерів: {full_containers}")
print(f"Залишилося одиниць: {leftover_items}")
```
